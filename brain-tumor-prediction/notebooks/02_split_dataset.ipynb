```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#**2. Load Dataset**

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/My Drive/Machine-Learning-Project/Brain-Tumor-Prediction/raw/Brain_Tumor_Prediction_Dataset.csv')

TARGET = 'Brain_Tumor_Present'
```

#**3. Split Dataset**

```python
from sklearn.model_selection import train_test_split

df_clean = df.dropna(subset=[TARGET]).reset_index(drop=True)

x = df_clean.drop(columns=[TARGET])
y = df_clean[TARGET]

x_train_val, x_test, y_train_val, y_test = train_test_split(
    x,y, test_size=0.10, random_state=42, stratify=y
)

x_train, x_val, y_train, y_val = train_test_split(
    x_train_val, y_train_val,
    test_size=0.20/0.90,
    random_state=42, stratify=y_train_val
)
```

Kemudian baris-baris dataset dipecah menjadi tiga bagian, yaitu data latih, data validasi, dan data uji dengan komposisi perbandingan 70:20:10 yang saya rasa cukup ideal untuk data sebanyak 250.000 baris.

Adapun alasan dilakukannya split dataset di tahap awal(sebelum preprocessing) adalah agar proses preprocessing seperti scaling, encoding, dan imputasi hanya di-fit pada data latih,  atau dengan kata lain agar tidak terjadi data leakage pada data uji (`x_test` dan `y_test`). Sehingga memecah dataset di tahap awal merupakan suatu solusi yang cukup efektif.
